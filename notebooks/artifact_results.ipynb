# Published Hansard Results (Artifact-Only)

This notebook loads the derived CSV artifacts committed in `data/`. It does not require the raw Hansard extract and cannot reproduce analyses that depend on individual interventions, payloads, or speaker-level raw records.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

# Resolve data/ whether the notebook is launched from the repository root
# or from the notebooks/ directory.
DATA_DIR = Path('data')
if not DATA_DIR.exists():
    DATA_DIR = Path('..') / 'data'

ARTIFACTS = {
    'monthly_attention_concepts': 'monthly_attention_concepts.csv',
    'monthly_tariff_topics': 'monthly_tariff_topics.csv',
    'member_tariff_records': 'member_tariff_records.csv',
    'topic_share': 'topic_share_2022_2026.csv',
    'historic_vs_last4q': 'historic_vs_last4q.csv',
}

missing = [name for name, filename in ARTIFACTS.items()
          if not (DATA_DIR / filename).exists()]
if missing:
    raise FileNotFoundError(f'Missing committed artifact(s): {missing}')

artifacts = {name: pd.read_csv(DATA_DIR / filename)
             for name, filename in ARTIFACTS.items()}
globals().update(artifacts)

print(f'Loaded {len(artifacts)} derived artifacts from {DATA_DIR.resolve()}')
for name, frame in artifacts.items():
    print(f'{name:28s} {frame.shape[0]:>8,} rows x {frame.shape[1]:>2} columns')
```
